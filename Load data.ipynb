```python
import pandas as pd

import numpy as np

import mysql.connector as con
```

```python
# Load the dataframe

df = pd.read_csv(r"d:\my multiple projects\end to end ecomerce sales project\ecommerce_cleaned_dataset.csv")

# Convert NaN values to Python None for MySQL NULL

df = df.astype(object)

df = df.where(pd.notna(df), None)

df.shape
```

```python
# Make connection

mycon = con.connect(
    host="localhost",
    user="root",
    password="1234",
    database="ecommerce_sales_data"
)


cursor = mycon.cursor()
```

```python
# Verify connection

if mycon.is_connected():
    print("Connected successfully")
else:
    print("Connection failed")
```

```python
# Insert query

insert_query = """
INSERT INTO ecommerce_raw
(
order_id,
customer_id,
customer_name,
gender,
age,
email,
phone_number,
country,
state,
city,
postal_code,
registration_date,
order_date,
shipping_date,
delivery_date,
product_id,
sku,
product_name,
category,
sub_category,
brand,
unit_price,
quantity,
discount_percentage,
discount_amount,
tax,
shipping_cost,
cost,
profit,
total_amount,
payment_method,
payment_status,
order_status,
delivery_method,
warehouse,
seller_id,
seller_name,
coupon_code,
returned,
return_reason,
refund_amount,
customer_rating,
review_count,
customer_segment,
device_type,
traffic_source,
session_id,
currency,
sales_channel
)
VALUES
(
%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,
%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,
%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,
%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,
%s,%s,%s,%s,%s,%s,%s,%s,%s
)
"""
```

```python
# Convert dataframe into records

data = [tuple(row) for row in df.values]
```

```python
# Load data to mysql in batches

batch_size = 5000

for i in range(0, len(data), batch_size):

    batch = data[i:i+batch_size]

    cursor.executemany(
        insert_query,
        batch
    )

    mycon.commit()

    print(f"{i + len(batch)} rows inserted")
```
