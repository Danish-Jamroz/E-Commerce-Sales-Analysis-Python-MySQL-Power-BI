Handle missing records


```python
# Detect missing records
df.isnull().sum()
```

```python
# Fill missing records in phone number

df["phone_number"] = df["phone_number"].fillna("Not Provided")
```

```python
# Fill missing records in the discount percentage column

original_value = df["quantity"] * df["unit_price"]

discount_percentage = (df["discount_amount"] / original_value) * 100

mask = (
    df["discount_percentage"].isna()
    
    & (original_value > 0)
    
    & (df["discount_amount"].notna())
)

# Update only missing discount percentage values

df.loc[mask, "discount_percentage"] = discount_percentage[mask]
```

```python
#fill missing values in the seller name

# Create mapping from existing seller records
seller_mapping = (
    
    df.dropna(subset=["seller_name"])
      
      .drop_duplicates("seller_id")
      
      .set_index("seller_id")["seller_name"]
)

# Fill missing seller_name using seller_id lookup

df["seller_name"] = df["seller_name"].fillna(df["seller_id"].map(seller_mapping))
```

```python
# Fill missing values in return reason 

df["returned"] = (
    df["returned"]
    .str.strip()
    .str.title()
)

mask = (
        
        (df["returned"] == "Yes")
        
        & (df["return_reason"].isna())
        
)

df.loc[mask,"return_reason"] = "Unknown"


mask = (
        
        (df["returned"] == "No")
        
        & (df["return_reason"].isna())
        
)

df.loc[mask,"return_reason"] = "Not Returned"
  
```

```python
# Fill missing records in the Review count colunm

df["review_count"] = df["review_count"].fillna(0)
```

```python
# Fill missing records in device_type, trafic_source and session_id column

columns = ["device_type","traffic_source","session_id"]

for col in columns:
    
    df[col] = df[col].fillna("Unknown")
```

Handle Outliers

```python
# Make function to detect outliers

def detect_outlier(x):
    
   Q1 = df[x].quantile(0.25)
   
   Q3 = df[x].quantile(0.75)
   
   IQR = Q3 - Q1
   
   Lower_bound = Q1 - (1.5 * IQR)
   
   Upper_bound = Q3 + (1.5 * IQR)
   
   outlier = df[(df[x] < Lower_bound) | (df[x] > Upper_bound)]
   
   return outlier

# Make function to remove outlier

def remove_outlier(x):
    
   Q1 = df[x].quantile(0.25)
   
   Q3 = df[x].quantile(0.75)
   
   IQR = Q3 - Q1
   
   Lower_bound = Q1 - (1.5 * IQR)
   
   Upper_bound = Q3 + (1.5 * IQR)
   
   clean_df = df[(df[x] >= Lower_bound) & (df[x] <= Upper_bound)]
   
   return clean_df
   
```
